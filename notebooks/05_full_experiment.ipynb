If it stops, run the same command again: finished runs are skipped. This notebook only watches progress.

````python
import os
from pathlib import Path
import pandas as pd
from IPython.display import display
from hpo import config, data, spaces, starts, methods, runner, stats, plots

pd.set_option("display.precision", 5)
# Results folder: the full run by default. Set HPO_RESULTS=smoke to read the smoke test.
RESULTS = (
    config.SMOKE_DIR if os.environ.get("HPO_RESULTS") == "smoke" else config.RESULTS_DIR
)
FIGS, TABLES = RESULTS / "figures", RESULTS / "tables"
TABLES.mkdir(parents=True, exist_ok=True)
print("Reading/writing:", RESULTS.relative_to(config.ROOT))
````

````python
runs_path = RESULTS / "runs.csv"
if runs_path.exists():
    runs, _ = stats.load(RESULTS)
    chk = stats.check_runs(runs)
    total = len(config.MODELS) * len(config.METHODS) * len(config.SEEDS)
    print(
        f"{chk['runs']}/{total} runs done | all full budget: {chk['all_full_budget']} | duplicates: {chk['duplicates']}"
    )
    display(runs.groupby(["model", "method"]).size().unstack())
else:
    print("No runs yet - start python -m hpo.run_all")
````
